```python
# plot histogram of naive samples
import matplotlib.pyplot as plt
import numpy as np
plt.hist(data['cd'][1].flatten(), bins=100, density=True)
lam = 1 
potential_fn = lambda q: (1-lam)*np.sum(0.5*(q**2)) + lam*np.sum(((q**2-5))**2)
# plot a line for the potential, using quadrature to normalize
x_grid = np.linspace(-5, 5, 1000)
rho = np.array([np.exp(-potential_fn(x)) for x in x_grid])
rho = rho / np.trapz(rho, x_grid)  # Normalize
plt.plot(x_grid, rho, 'k--', linewidth=2, label='True distribution')

# data['naive'][1].shape
```
